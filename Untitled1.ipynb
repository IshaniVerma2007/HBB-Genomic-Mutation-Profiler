```python
# DNA sequences for a snippet of the Human HBB Gene (Exon 1 region)
healthy_dna = "ATGGTGCACCTGACTCCTGAGGAGAAGTCTGCCGTTACTGCCCTGTGGGGCAAGGTG"
mutated_dna = "ATGGTGCACCTGACTCCTGTGGAGAAGTCTGCCGTTACTGCCCTGTGGGGCAAGGTG"

print("✓ Genetic sequences loaded successfully.")
```

```python
from Bio.Seq import Seq

# 1. Convert raw text strings into formal BioPython Sequence Objects
seq_normal = Seq(healthy_dna)
seq_mutant = Seq(mutated_dna)

# 2. Programmatically scan for discrepancies
mutation_found = False
for index in range(len(seq_normal)):
    if seq_normal[index] != seq_mutant[index]:
        print(f"🚨 Mutation Found at Nucleotide Position {index + 1}!")
        print(f"Healthy Base: {seq_normal[index]} ➔ Patient Mutant Base: {seq_mutant[index]}")
        mutation_found = True

if not mutation_found:
    print("Sequences are completely identical.")
```

```python
# Translate the DNA sequence directly into an Amino Acid sequence
protein_normal = seq_normal.translate()
protein_mutant = seq_mutant.translate()

print("Healthy Peptide Chain: ", protein_normal)
print("Mutated Peptide Chain: ", protein_mutant)

# Find the exact amino acid shift
for idx in range(len(protein_normal)):
    if protein_normal[idx] != protein_mutant[idx]:
        print(f"\n💡 Molecular Consequence: Amino Acid change at residue position {idx + 1}!")
        print(f"Normal Residue: {protein_normal[idx]} (Glutamic Acid) ➔ Mutant Residue: {protein_mutant[idx]} (Valine)")
```
